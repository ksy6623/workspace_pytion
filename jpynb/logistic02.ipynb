```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
df = pd.read_csv(r"C:\dev\workspace_pytion\dataset\Titanic Passengers.csv")
df.shape
```

```python
# 1: 생존, 0 : 사망
# 1: 여자, 0 : 남자
df['sex'] = df['sex'].map({'female':1,'male':0})
# age 평균
df['age'] = df['age'].fillna(value=df['age'].mean())
# first :1, business 2, economy 3 을 0 or 1로
# pandas apply로 모든 행에 사용자 함수를 적용할 수 있음.
df['first'] = df['pclass'].apply(lambda x : 1 if x == 1 else 0)
df['business'] = df['pclass'].apply(lambda x : 1 if x == 2 else 0)
df['economy'] = df['pclass'].apply(lambda x : 1 if x == 3 else 0)
x = df[['sex','age','first','business','economy']]
y = df['survived']
x.head()
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
model = LogisticRegression()
model.fit(x_train, y_train)
print('기울기:',model.coef_)
print('y절편:',model.intercept_)
print('테스트 성능:',model.score(x_test,y_test))
```
