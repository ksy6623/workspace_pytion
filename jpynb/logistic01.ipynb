```python
import numpy as np
import matplotlib.pyplot as plt
# x:공부시간, y:(합격[1]/불합격[0])
x = np.array([2,4,6,8,10,12,14])
y = np.array([0,0,0,1,1,1,1])

def sigmoid(p_x):
    return 1 / (1+ np.exp(-p_x))
```

```python
test_x = np.linspace(-10,10,400) # -10 ~10 랜덤값
test_y = sigmoid(test_x)
print(test_x)
print(test_y)
```

```python
plt.figure(figsize=(8,6))
plt.plot(test_x,test_y)
plt.xlabel("x")
plt.grid(True)
plt.legend()
```

```python
a = 0 # 기울기
b = 0 # y절편
lr = 0.05 # 학습률
epochs = 2001
plt.scatter(x,y)
n = len(x)
for i in range(epochs):
    y_pred = sigmoid(a * x + b) # 예측값
    error = y - y_pred
    a_diff = -(1/n) * sum(x * error) #a에 대한 편미분
    b_diff = -(1/n) * sum(error)
    a = a - lr * a_diff
    b = b - lr * b_diff
    if i % 100 == 0:
        print(f"epoch={i}, a={a},b={b}")
x_range = np.arange(0,15,0.1)
plt.plot(x_range, sigmoid(a * x_range + b))
plt.show()
```
